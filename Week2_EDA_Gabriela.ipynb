```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python

url = "https://raw.githubusercontent.com/joaoadijami-hub/NextGenLearners-DataScience-Week1/main/applicants_cleaned.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)
```

```python

print("Column names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())
```


## 1. Which Domain Received the Most Applications?

This analysis compares the number of applications received by each domain.


```python

domain_counts = df['Domain Applied'].value_counts()

print(domain_counts)

plt.figure(figsize=(8, 5))
domain_counts.plot(kind='bar')

plt.title('Number of Applications per Domain')
plt.xlabel('Domain')
plt.ylabel('Number of Applications')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```


## 2. How Did Applications Trend Over Time?

This analysis examines how application volume changed from month to month.


```python

df['Application Date'] = pd.to_datetime(
    df['Application Date'],
    errors='coerce'
)

df['Application Month'] = (
    df['Application Date'].dt.to_period('M')
)

monthly_trend = df.groupby('Application Month').size()

print(monthly_trend)

plt.figure(figsize=(8, 5))
monthly_trend.plot(kind='line', marker='o')

plt.title('Application Trend Over Time (Monthly)')
plt.xlabel('Month')
plt.ylabel('Number of Applications')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```


## 3. What Is the Overall Application Status Breakdown?

This analysis compares the number of selected, rejected, and pending applications.


```python

status_counts = df['Status'].value_counts()

print(status_counts)

plt.figure(figsize=(7, 7))
status_counts.plot(kind='pie', autopct='%1.1f%%')

plt.title('Application Status Distribution')
plt.ylabel('')
plt.tight_layout()
plt.show()
```


## 4. What Is the Acceptance Rate by Domain?

This analysis compares selection rates across domains rather than comparing application counts alone.


```python

df['Status Clean'] = df['Status'].astype(str).str.strip().str.lower()

df['Selected Flag'] = (
    df['Status Clean'] == 'selected'
).astype(int)

acceptance_table = df.pivot_table(
    index='Domain Applied',
    values='Selected Flag',
    aggfunc='mean'
) * 100

acceptance_table.columns = ['Acceptance Rate (%)']

print(acceptance_table)

acceptance_table.plot(
    kind='bar',
    figsize=(8, 5),
    legend=False
)

plt.title('Acceptance Rate by Domain')
plt.xlabel('Domain')
plt.ylabel('Acceptance Rate (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```


## 5. Which Universities Have the Most Applicants?

This analysis identifies the universities contributing the most applications.


```python

university_counts = df['University'].value_counts().head(10)

print(university_counts)

plt.figure(figsize=(9, 6))
university_counts.sort_values().plot(kind='barh')

plt.title('Top 10 Universities by Number of Applications')
plt.xlabel('Number of Applications')
plt.ylabel('University')
plt.tight_layout()
plt.show()
```


## 6. Bonus Analysis: Application Status by Domain

This chart compares application outcomes across domains to identify differences in their status distributions.


```python

status_by_domain = pd.crosstab(
    df['Domain Applied'],
    df['Status']
)

print(status_by_domain)

status_by_domain.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Application Status by Domain')
plt.xlabel('Domain')
plt.ylabel('Number of Applications')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Status')
plt.tight_layout()
plt.show()
```
