## Objective

The objective of this project is to clean and explore an applicant dataset using Python and Pandas. The process includes identifying missing values, removing duplicate records, standardizing text, and correcting data types.

```python

import pandas as pd
```

```python

import pandas as pd

# Create a fictional applicant dataset
data = {
    "Applicant_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 102, 112],
    "Name": [
        "Ana Silva", "Bruno Costa", "Maria Santos", "Pedro Manuel",
        "Carla Mendes", "Joao Paulo", "Sofia Lima", "Daniel Alves",
        "Beatriz Gomes", "Lucas Pereira", "Bruno Costa", "Ines Rocha"
    ],
    "Age": [24, 29, None, 35, 27, 22, 31, None, 26, 28, 29, 24],
    "Gender": [
        "Female", "Male", "FEMALE", "Male", "female", "Male",
        "Female", "MALE", "Female", "Male", "Male", "Female"
    ],
    "Education": [
        "Bachelor", "Master", "Bachelor", "PhD", "Bachelor",
        "High School", "Master", "Bachelor", "Master", "Bachelor",
        "Master", "Bachelor"
    ],
    "Years_Experience": [2, 5, 3, 10, None, 1, 7, 4, 2, 6, 5, 0],
    "Applied_Position": [
        "Data Analyst", "Software Engineer", "Data Analyst",
        "Data Scientist", "Business Analyst", "Intern",
        "Data Analyst", "Software Engineer", "Business Analyst",
        "Data Scientist", "Software Engineer", "Intern"
    ],
    "Application_Status": [
        "Accepted", "Rejected", "Under Review", "Accepted",
        "Rejected", "under review", "Accepted", "Rejected",
        "Accepted", "Under Review", "Rejected", "Accepted"
    ],
    "Expected_Salary": [
        2500, 4000, 2800, 6000, None, 1200,
        3500, 4200, 3000, 5000, 4000, 1000
    ]
}

df = pd.DataFrame(data)

# Save the fictional dataset as a CSV file
df.to_csv("applicants.csv", index=False)

print("Dataset created successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python


print("Dataset dimensions:", df.shape)


df.info()


print("\nMissing values per column:")
print(df.isnull().sum())


print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python

print(df.isnull().sum())
```

```python


print("Completely duplicated rows:")
print(df.duplicated().sum())


print("\nRepeated Applicant IDs:")
print(df[df["Applicant_ID"].duplicated(keep=False)].sort_values("Applicant_ID"))
```

```python


df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)

print("Remaining duplicate rows:", df.duplicated().sum())
```

```python

print("Median age:", df["Age"].median())
print("Median years of experience:", df["Years_Experience"].median())
print("Median expected salary:", df["Expected_Salary"].median())
```

```python

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Years_Experience"] = df["Years_Experience"].fillna(
    df["Years_Experience"].median()
)
df["Expected_Salary"] = df["Expected_Salary"].fillna(
    df["Expected_Salary"].median()
)


print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python

print("Gender categories:")
print(df["Gender"].unique())

print("\nApplication status categories:")
print(df["Application_Status"].unique())
```

```python

df["Gender"] = df["Gender"].str.strip().str.title()

df["Application_Status"] = (
    df["Application_Status"].str.strip().str.title()
)


print("Gender categories after cleaning:")
print(df["Gender"].unique())

print("\nApplication status categories after cleaning:")
print(df["Application_Status"].unique())
```

```python

print(df.dtypes)

#
df["Age"] = df["Age"].astype(int)
df["Years_Experience"] = df["Years_Experience"].astype(int)


print(df.dtypes)

#

print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nFirst five rows:")
display(df.head())

```

```python



print("Number of rows and columns:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nUnique applicant IDs:", df["Applicant_ID"].nunique())

print("\nDataset preview:")
display(df.head())


# Save the cleaned dataset
df.to_csv("applicants_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
